# **2. Decidir como se va realizar la evaluación**

El objetivo de esta sección es establecer una estrategia de evaluación fiable que nos permita:

- Comparar diferentes algoritmos y configuraciones (evaluación interna, o *inner*).
- Estimar el rendimiento futuro del modelo final en datos nuevos (evaluación externa, o *outer*).

Dado que se trata de un problema de clasificación **desbalanceado** (la variable objetivo `Attrition` no está equilibrada entre clases), hemos decidido usar métricas más apropiadas que el simple `accuracy`, como:

- **Balanced Accuracy**: media de TPR (True Positive Rate) y TNR (True Negative Rate).
- **TPR**: proporción de positivos correctamente clasificados.
- **TNR**: proporción de negativos correctamente clasificados.
- **Matriz de confusión**: para visualizar los aciertos y errores por clase.

La evaluación externa (*outer*) solo se utilizará **una vez**, al final del proceso, para estimar el rendimiento futuro. El resto de decisiones (modelo, escalado, hiperparámetros, etc.) se tomarán exclusivamente con evaluación interna (*inner*) sobre el conjunto de entrenamiento.

## Evaluación Outer, División en train/test

```python
df = pd.read_csv('/content/attrition_availabledata_10.csv.gz')

from sklearn.model_selection import train_test_split

# Variable objetivo y atributos
X = df.drop(columns=["Attrition"])  # Asegúrate de que 'Attrition' es tu variable objetivo
y = df["Attrition"]

# División 2/3 - 1/3, estratificada para mantener proporción de clases
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=1/3,
    stratify=y,
    random_state=42
)

print("Tamaño conjunto entrenamiento:", X_train.shape)
print("Tamaño conjunto test:", X_test.shape)
```

Se ha realizado una partición estratificada de los datos en 2/3 para entrenamiento y 1/3 para test. Esta evaluación solo se usará **una vez al final**, después de haber elegido el mejor modelo con evaluación interna. De esta manera, garantizamos una estimación honesta del rendimiento del modelo final sobre datos nuevos.


## Evaluación Inner  con validación cruzada

```python
from sklearn.model_selection import StratifiedKFold

# Definición de la validación cruzada (5 folds, estratificada)
cv_inner = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

Durante el desarrollo del modelo, se utilizará validación cruzada estratificada (5-fold). Esto permite:

## Métricas elegidas

Como el problema presenta un desbalance de clases, no utilizamos `accuracy` como métrica principal. En su lugar, empleamos:

- `Balanced Accuracy`: más adecuada en este contexto, ya que exige buen rendimiento tanto para positivos como para negativos.

# **3. Métodos Básicos: KNN Y Trees**

## Selección de preprocesamiento


KNN es un algoritmo sensible a la escala y al tratamiento de valores faltantes, por eso necesitamos probar distintos preprocesamientos. Antes de aplicar estos modelos, probamos combinaciones de escalado (MinMax, Standard, Robust) e imputación (media, mediana) usando KNN. El objetivo era identificar el preprocesamiento que proporcionara mejor rendimiento (balanced accuracy).

Vamos a probar: \
**- 3 escalados:**

MinMaxScaler

StandardScaler

RobustScaler

**- 2 imputaciones:**

media

mediana

```python
import time
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import make_scorer, balanced_accuracy_score

# Columnas categóricas y numéricas
cat_cols = X_train.select_dtypes(include='object').columns.tolist()
num_cols = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Estrategias a probar
scalers = {
    "minmax": MinMaxScaler(),
    "standard": StandardScaler(),
    "robust": RobustScaler()
}

imputers = {
    "media": SimpleImputer(strategy="mean"),
    "mediana": SimpleImputer(strategy="median")
}

# Validación cruzada
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Almacenar resultados
resultados = []

for nombre_scaler, scaler in scalers.items():
    for nombre_imputer, imputer in imputers.items():
        # Preprocesamiento numérico
        num_pipeline = Pipeline([
            ("imputer", imputer),
            ("scaler", scaler)
        ])

        # Preprocesamiento categórico
        cat_pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ])

        # Combinación
        preprocessor = ColumnTransformer([
            ("num", num_pipeline, num_cols),
            ("cat", cat_pipeline, cat_cols)
        ])

        # Pipeline completo
        knn_pipeline = Pipeline([
            ("preprocessor", preprocessor),
            ("model", KNeighborsClassifier())
        ])

        # Medir tiempo y score
        start = time.time()
        score = cross_val_score(
            knn_pipeline, X_train, y_train,
            cv=cv, scoring=make_scorer(balanced_accuracy_score)
        )
        end = time.time()

        resultados.append({
            "escalado": nombre_scaler,
            "imputación": nombre_imputer,
            "balanced_accuracy": score.mean(),
            "tiempo_seg": round(end - start, 2)
        })

# Mostrar resultados ordenados
import pandas as pd
pd.DataFrame(resultados).sort_values("balanced_accuracy", ascending=False)
```

Hemos obtenido como resultado que la mejor combinación de posibles escaladores e imputaciones es utilizando minmax y media, con un balanced_accuracy de 0.612

## Preprocesamiento común

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler

# Columnas numéricas y categóricas
num_cols = X_train.select_dtypes(include=["int64", "float64"]).columns.tolist()
cat_cols = X_train.select_dtypes(include=["object"]).columns.tolist()

# Pipeline numérico (media + minmax)
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", MinMaxScaler())
])

# Pipeline categórico (más frecuente + one hot)
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combinación en ColumnTransformer
preprocessor = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])
```

##Evaluación con hiperparámetros por defecto

Se evalúan ambos modelos usando validación cruzada estratificada (5 folds) y se mide el tiempo de entrenamiento. La métrica principal es `balanced_accuracy`.

Esto nos permite observar el rendimiento inicial de cada modelo antes de realizar ajuste de hiperparámetros.


```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import make_scorer, balanced_accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
import time

# Validación cruzada
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Pipelines
pipeline_knn = Pipeline([
    ("preprocessor", preprocessor),
    ("model", KNeighborsClassifier())
])

pipeline_tree = Pipeline([
    ("preprocessor", preprocessor),
    ("model", DecisionTreeClassifier(random_state=42))
])

# Función para evaluar modelos
def evaluar_pipeline(pipeline, nombre):
    start = time.time()
    scores = cross_val_score(pipeline, X_train, y_train, cv=cv, scoring=make_scorer(balanced_accuracy_score))
    end = time.time()
    print(f"{nombre}:")
    print(f"  Balanced Accuracy media: {scores.mean():.4f}")
    print(f"  Tiempo de entrenamiento: {end - start:.2f} segundos\n")

evaluar_pipeline(pipeline_knn, "KNN (default)")
evaluar_pipeline(pipeline_tree, "Árbol (default)")
```

## Ajuste de hiperparámetros (HPO)

A continuación se ajustan los hiperparámetros más importantes de cada modelo:

- `KNN`: número de vecinos (`n_neighbors`)
- `Árbol`: profundidad máxima del árbol (`max_depth`)

Se utiliza `GridSearchCV` con validación cruzada estratificada.


```python
from sklearn.model_selection import GridSearchCV

# KNN
param_knn = {"model__n_neighbors": range(1, 21)}
grid_knn = GridSearchCV(pipeline_knn, param_grid=param_knn, cv=cv, scoring="balanced_accuracy", n_jobs=-1)
grid_knn.fit(X_train, y_train)

# Árbol
param_tree = {"model__max_depth": range(2, 21)}
grid_tree = GridSearchCV(pipeline_tree, param_grid=param_tree, cv=cv, scoring="balanced_accuracy", n_jobs=-1)
grid_tree.fit(X_train, y_train)

# Resultados
print("Mejor KNN:", grid_knn.best_params_, "Balanced Accuracy:", grid_knn.best_score_)
print("Mejor Árbol:", grid_tree.best_params_, "Balanced Accuracy:", grid_tree.best_score_)
```

Se muestra cómo afecta el número de vecinos en KNN y la profundidad máxima en los árboles de decisión sobre la métrica de `balanced_accuracy`.


```python
import matplotlib.pyplot as plt

# KNN
knn_scores = grid_knn.cv_results_["mean_test_score"]
knn_k = param_knn["model__n_neighbors"]
plt.plot(knn_k, knn_scores, marker='o')
plt.title("KNN: efecto de n_neighbors")
plt.xlabel("n_neighbors")
plt.ylabel("Balanced Accuracy")
plt.grid(True)
plt.show()

# Árbol
tree_scores = grid_tree.cv_results_["mean_test_score"]
tree_depths = param_tree["model__max_depth"]
plt.plot(tree_depths, tree_scores, marker='o', color='green')
plt.title("Árbol: efecto de max_depth")
plt.xlabel("max_depth")
plt.ylabel("Balanced Accuracy")
plt.grid(True)
plt.show()
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import balanced_accuracy_score, accuracy_score, recall_score, confusion_matrix
from sklearn.model_selection import cross_val_score

# DummyClassifier que predice siempre la clase más frecuente
dummy_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", DummyClassifier(strategy="most_frequent"))
])

# Validación cruzada como en los otros modelos
dummy_scores = cross_val_score(
    dummy_pipeline, X_train, y_train,
    cv=cv,
    scoring=make_scorer(balanced_accuracy_score)
)

print("DummyClassifier (estrategia 'most_frequent'):")
print(f"  Balanced Accuracy media: {dummy_scores.mean():.4f}")
```
