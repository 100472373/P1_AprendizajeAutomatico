## Variables e instancias

```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('/content/attrition_competition_10.csv.gz')
```

```python
# Número de variables e instancias
num_variables = df.shape[1]
num_instancias = df.shape[0]
print(f"El conjunto de datos tiene {num_instancias} instancias y {num_variables} variables.")
```

```python
df.describe()
```

## Variables categóricas, ordinales y numéricas

```python
# Verificar los tipos de datos
print("Tipos de datos de las variables:")
print(df.dtypes)

# Variables categóricas: aquellas que son de tipo 'object' o 'category'
categorical_vars = df.select_dtypes(include=['object']).columns

# Variables numéricas: aquellas que son de tipo 'int' o 'float'
numerical_vars = df.select_dtypes(include=['int64', 'float64']).columns

# Variables ordinales: se pueden identificar por la naturaleza de los datos, pero como simplificación se pueden incluir las numéricas con escalas de valor finitas (ejemplo: JobSatisfaction)
ordinal_vars = ['JobInvolvement', 'PerformanceRating', 'EnvironmentSatisfaction', 'JobSatisfaction', 'WorkLifeBalance']

print(f"\nVariables categóricas: {categorical_vars}")
print(f"Variables numéricas: {numerical_vars}")
print(f"Variables ordinales: {ordinal_vars}")
```

## Alta cardinalidad, valores faltantes y columnas constantes

```python
# Identificar variables con alta cardinalidad
high_cardinality_vars = [col for col in categorical_vars if df[col].nunique() > 10]
print(f"\nVariables con alta cardinalidad: {high_cardinality_vars}")
```

```python
# Verificar valores faltantes
missing_values = df.isnull().sum()

# Mostrar columnas con valores faltantes
missing_data = missing_values[missing_values > 0]
print(f"\nValores faltantes:\n{missing_data}")
```

```python
# Identificar columnas constantes o de ID
constant_vars = df.columns[df.nunique() == 1]
print(f"\nVariables constantes: {constant_vars}")

# Identificar columnas de ID (suponiendo que 'EmployeeID' es un ID único)
id_vars = ['EmployeeID']
print(f"Variables de ID: {id_vars}")
```
