# **2. Decidir como se va realizar la evaluación**

El objetivo de esta sección es establecer una estrategia de evaluación fiable que nos permita:

- Comparar diferentes algoritmos y configuraciones (evaluación interna o *inner*).
- Estimar el rendimiento futuro del modelo final en datos nuevos (evaluación externa o *outer*).

Dado que se trata de un problema de clasificación **desbalanceado** (la variable objetivo `Attrition` no está equilibrada entre clases), hemos decidido usar métricas mejores, para este caso, que el simple `accuracy`, como:

- **Balanced Accuracy**: media de TPR (True Positive Rate) y TNR (True Negative Rate).
- **TPR**: proporción de positivos correctamente clasificados.
- **TNR**: proporción de negativos correctamente clasificados.
- **Matriz de confusión**: para visualizar los aciertos y errores por clase.

La evaluación externa (*outer*) solo se utilizará **una vez**, al final del proceso, para estimar el rendimiento futuro. El resto de decisiones (modelo, escalado, hiperparámetros, etc.) se realizarán exclusivamente con evaluación interna (*inner*) sobre el conjunto de entrenamiento.

## Evaluación Outer, División en train/test

```python
df = pd.read_csv('/content/attrition_availabledata_10.csv.gz')

from sklearn.model_selection import train_test_split

# Variable objetivo y atributos
X = df.drop(columns=["Attrition"])
y = df["Attrition"]

# División 2/3 - 1/3
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=1/3,
    stratify=y,
    random_state=42
)

print("Conjunto entrenamiento:", X_train.shape)
print("Conjunto test:", X_test.shape)
```

Se ha realizado una partición estratificada de los datos en 2/3 para entrenamiento y 1/3 para test. Esta evaluación solo se usará **una vez al final**, después de haber elegido el mejor modelo con evaluación interna. De esta manera, garantizamos una estimación real del rendimiento del modelo final sobre datos nuevos.

## Selección de preprocesamiento


KNN es un algoritmo sensible a la escala y al tratamiento de valores faltantes, por eso necesitamos probar distintos preprocesamientos. Antes de aplicar estos modelos, probamos combinaciones de escalado (MinMax, Standard, Robust) e imputación (media, mediana) usando KNN. El objetivo es identificar el preprocesamiento que proporcionara mejor rendimiento (balanced accuracy).

Vamos a probar: \
**- 3 escalados:**

MinMaxScaler

StandardScaler

RobustScaler

**- 2 imputaciones:**

media

mediana

```python
import time
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import make_scorer, balanced_accuracy_score

# Columnas categóricas y numéricas
colums_cat = X_train.select_dtypes(include='object').columns.tolist()
colums_num = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Combinaciones a probar
scalers = {
    "minmax": MinMaxScaler(),
    "standard": StandardScaler(),
    "robust": RobustScaler()
}

imputers = {
    "media": SimpleImputer(strategy="mean"),
    "mediana": SimpleImputer(strategy="median")
}

# Validación cruzada
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

# Almacenar resultados
resultados = []

for nombre_scaler, scaler in scalers.items():
    for nombre_imputer, imputer in imputers.items():
        # Preprocesamiento numérico
        num_pipeline = Pipeline([
            ("imputer", imputer),
            ("scaler", scaler)
        ])

        # Preprocesamiento categórico
        cat_pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ])

        # Combinación
        preprocessor = ColumnTransformer([
            ("num", num_pipeline, colums_num),
            ("cat", cat_pipeline, colums_cat)
        ])

        # Pipeline completo
        knn_pipeline = Pipeline([
            ("preprocessor", preprocessor),
            ("model", KNeighborsClassifier())
        ])

        # Medir tiempo y score
        comienzo = time.time()
        score = cross_val_score(
            knn_pipeline, X_train, y_train,
            cv=cv, scoring=make_scorer(balanced_accuracy_score)
        )
        fin = time.time()

        resultados.append({
            "escalado": nombre_scaler,
            "imputación": nombre_imputer,
            "balanced_accuracy": score.mean(),
            "tiempo_seg": round(fin - comienzo, 2)
        })

# Mostrar resultados ordenados
import pandas as pd
pd.DataFrame(resultados).sort_values("balanced_accuracy", ascending=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Creamos DataFrame a partir de resultados
df_resultados = pd.DataFrame(resultados)

# Combinamos escalado + imputación en una sola etiqueta para el gráfico
df_resultados["combinación"] = df_resultados["escalado"] + " + " + df_resultados["imputación"]

# Ordenamos por balanced_accuracy
df_resultados = df_resultados.sort_values("balanced_accuracy", ascending=False)

# Gráfico de barras de balanced_accuracy
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df_resultados,
    x="balanced_accuracy",
    y="combinación",
    palette="viridis"
)
plt.xlabel("Balanced Accuracy")
plt.ylabel("Combinación de Escalado + Imputación")
plt.title("Comparación de Preprocesamientos con KNN")
plt.xlim(0.5, 0.7)  # ajusta según tus resultados
plt.grid(axis="x", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```



Hemos obtenido como resultado que la mejor combinación de posibles escaladores e imputaciones es utilizando `robustScaler y media`, con un balanced_accuracy de `0.5972`





## Preprocesamiento común

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, RobustScaler

# Columnas numéricas y categóricas
num_cols = X_train.select_dtypes(include=["int64", "float64"]).columns.tolist()
cat_cols = X_train.select_dtypes(include=["object"]).columns.tolist()

# Pipeline numérico (media + robust)
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", RobustScaler())
])

# Pipeline categórico (más frecuente + one hot)
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combinación en ColumnTransformer
preprocessor = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])
```

##Evaluación con hiperparámetros por defecto

Se evalúan ambos modelos usando validación cruzada estratificada (3 folds) y se mide el tiempo de entrenamiento. La métrica principal es `balanced_accuracy`.

Esto nos permite observar el rendimiento inicial de cada modelo antes de realizar ajuste de hiperparámetros.


```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import make_scorer, balanced_accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
import time

# Validación cruzada
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

# Pipelines
pipeline_knn = Pipeline([
    ("preprocessor", preprocessor),
    ("model", KNeighborsClassifier())
])

pipeline_tree = Pipeline([
    ("preprocessor", preprocessor),
    ("model", DecisionTreeClassifier(random_state=42))
])

# Evalumos los modelos
def evaluar_pipeline(pipeline, nombre):
    start = time.time()
    scores = cross_val_score(pipeline, X_train, y_train, cv=cv, scoring=make_scorer(balanced_accuracy_score))
    end = time.time()
    print(f"{nombre}:")
    print(f"  Balanced Accuracy media: {scores.mean():.4f}")
    print(f"  Tiempo de entrenamiento: {end - start:.2f} s\n")

evaluar_pipeline(pipeline_knn, "KNN (default)")
evaluar_pipeline(pipeline_tree, "Árbol (default)")
```

## Ajuste de hiperparámetros (HPO)

A continuación se ajustan los hiperparámetros más importantes de cada modelo:

- `KNN`: número de vecinos (`n_neighbors`)
- `Árbol`: profundidad máxima del árbol (`max_depth`)

Utilizamos `GridSearchCV` con validación cruzada estratificada.


```python
from sklearn.model_selection import GridSearchCV

# KNN
param_knn = {"model__n_neighbors": range(1, 21)}
grid_knn = GridSearchCV(pipeline_knn, param_grid=param_knn, cv=cv, scoring="balanced_accuracy", n_jobs=-1)
grid_knn.fit(X_train, y_train)

# Árbol
param_tree = {"model__max_depth": range(2, 21)}
grid_tree = GridSearchCV(pipeline_tree, param_grid=param_tree, cv=cv, scoring="balanced_accuracy", n_jobs=-1)
grid_tree.fit(X_train, y_train)

# Resultados
print("Mejor KNN:", grid_knn.best_params_, "Balanced Accuracy:", grid_knn.best_score_)
print("Mejor Árbol:", grid_tree.best_params_, "Balanced Accuracy:", grid_tree.best_score_)
```

Se muestra cómo afecta el número de vecinos en KNN y la profundidad máxima en los árboles de decisión sobre la métrica de `balanced_accuracy`.


```python
import matplotlib.pyplot as plt

# KNN
knn_scores = grid_knn.cv_results_["mean_test_score"]
knn_k = param_knn["model__n_neighbors"]
plt.plot(knn_k, knn_scores, marker='o')
plt.title("KNN: efecto de n_neighbors")
plt.xlabel("n_neighbors")
plt.ylabel("Balanced Accuracy")
plt.grid(True)
plt.show()

# Árbol
tree_scores = grid_tree.cv_results_["mean_test_score"]
tree_depths = param_tree["model__max_depth"]
plt.plot(tree_depths, tree_scores, marker='o', color='green')
plt.title("Árbol: efecto de max_depth")
plt.xlabel("max_depth")
plt.ylabel("Balanced Accuracy")
plt.grid(True)
plt.show()
```

# **4. Avanzados: Modelos Lineales y SVMs**

En esta sección se aplican modelos más avanzados de clasificación:

- **Modelos lineales**: regresión logística sin y con regularización L1.
- **Support Vector Machines (SVM)**: modelos que buscan encontrar el hiperplano que mejor separa las clases.

Se evaluarán primero con sus hiperparámetros por defecto y luego se ajustarán los más relevantes mediante validación cruzada. Finalmente, se analizará si alguno de ellos permite interpretar la importancia de los atributos en el modelo.

## Preprocesamiento común

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
from sklearn.impute import SimpleImputer


# Columnas numéricas y categóricas
num_cols = X_train.select_dtypes(include=["int64", "float64"]).columns.tolist()
cat_cols = X_train.select_dtypes(include=["object"]).columns.tolist()

# Pipeline numérico (media + robust)
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", RobustScaler())
])

# Pipeline categórico (más frecuente + one hot)
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Preprocesador combinado
preprocessor = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])
```

## Evaluación con valores por defecto





```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import make_scorer, balanced_accuracy_score
import time

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

# Modelos
pipeline_logreg = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(penalty=None, max_iter=1000))
])


pipeline_l1 = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(penalty="l1", solver="liblinear", max_iter=1000))
])

pipeline_svm = Pipeline([
    ("preprocessor", preprocessor),
    ("model", SVC())
])

def evaluar_modelo(pipeline, nombre):
    comienzo = time.time()
    scores = cross_val_score(pipeline, X_train, y_train, cv=cv, scoring=make_scorer(balanced_accuracy_score))
    fin = time.time()
    print(f"{nombre}:")
    print(f"  Balanced Accuracy media: {scores.mean():.4f}")
    print(f"  Tiempo de entrenamiento: {fin - comienzo:.2f} segundos\n")

# Evaluar
evaluar_modelo(pipeline_logreg, "Regresión Logística (sin regularización)")
evaluar_modelo(pipeline_l1, "Regresión Logística (L1)")
evaluar_modelo(pipeline_svm, "SVM (default)")
```



Se evalúan los tres modelos usando validación cruzada estratificada con `balanced_accuracy` como métrica. Se mide el tiempo de entrenamiento para valorar el coste computacional.

> La regresión logística con regularización L1 establece una penalización que favorece la selección de atributos importantes, lo que será útil más adelante para interpretar el modelo.




## Ajuste de hiperparámetros (GridSearchCV)

```python
from sklearn.model_selection import GridSearchCV

# Hiperparámetros a ajustar
param_logreg_l1 = {
    "model__C": [0.01, 0.1, 1, 10, 100]
}

param_svm = {
    "model__C": [0.01, 0.1, 1, 10],
    "model__kernel": ["linear", "rbf"]
}

# Grid Search
grid_l1 = GridSearchCV(pipeline_l1, param_logreg_l1, cv=cv, scoring="balanced_accuracy", n_jobs=-1)
grid_svm = GridSearchCV(pipeline_svm, param_svm, cv=cv, scoring="balanced_accuracy", n_jobs=-1)

# Entrenamiento
comienzo = time.time()
grid_l1.fit(X_train, y_train)
fin = time.time()
print(f"Tiempo GridSearch L1: {fin - comienzo:.2f} segundos")

comienzo = time.time()
grid_svm.fit(X_train, y_train)
fin = time.time()
print(f"Tiempo GridSearch SVM: {fin - comienzo:.2f} segundos")

# Analizamos mejores resultados
print("Mejor LogReg L1:", grid_l1.best_params_, "→ BA:", grid_l1.best_score_)
print("Mejor SVM:", grid_svm.best_params_, "→ BA:", grid_svm.best_score_)
```

## ¿Qué modelo permite interpretar la importancia de los atributos?

La **regresión logística con regularización L1** (Lasso) fuerza algunos coeficientes a cero, eliminando atributos no relevantes. Esto permite identificar cuáles tienen un peso real en la decisión del modelo.

A continuación, se muestran los coeficientes de los atributos tras el entrenamiento.


```python
# Entrenamos para ver coeficientes
best_l1 = grid_l1.best_estimator_
best_l1.fit(X_train, y_train)

# Obtenemos los nombres de las variables tras one-hot encoding
ft_nombre_num = num_cols
ft_nombre_cat = best_l1.named_steps["preprocessor"].named_transformers_["cat"]["encoder"].get_feature_names_out(cat_cols)
feature_names = list(ft_nombre_num) + list(ft_nombre_cat)

# Coeficientes del modelo
import pandas as pd
coefs = best_l1.named_steps["model"].coef_[0]
coef_df = pd.DataFrame({
    "feature": feature_names,
    "coef": coefs
}).sort_values(by="coef", key=abs, ascending=False)

# Mostramos los más relevantes
coef_df[coef_df["coef"] != 0].head(10)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Filtramos los coeficientes diferentes de 0
coef_filtrados = coef_df[coef_df["coef"] != 0].copy()

# Ordenamos por importancia
coef_filtrados["abs_coef"] = np.abs(coef_filtrados["coef"])
coef_ordenados = coef_filtrados.sort_values("abs_coef", ascending=True)

# Gráfico de barras horizontales
plt.figure(figsize=(8, 6))
plt.barh(coef_ordenados["feature"], coef_ordenados["coef"], color="skyblue")
plt.axvline(0, color='gray', linestyle='--')
plt.title("Coeficientes del modelo (L1)")
plt.xlabel("Peso del atributo (coeficiente)")
plt.tight_layout()
plt.show()
```

### Análisis de coeficientes
Un coeficiente positivo indica que  es más probable es que la persona abandone la empresa.
Mientras que  coeficiente negativo indica que  es menos probable es que la persona abandone la empresa.

Atributos como **“llevar mucho tiempo con el jefe”** o **“no tener que viajar”** disminuyen la probabilidad de irse (`coeficiente negativo`).

En cambio, **“trabajar muchas horas”** aumenta la probabilidad de marcharse (`coeficiente positivo`).

Variables que aparentemente deberían tener una alta correlación con la variable solución en la vida real, como es el caso de `MonthlyIncome`, nos damos cuenta que el modelo la clasifica con un coeficiente practicamente nulo.

Esto puede ocurrir debido a que esta variable pueda estar fuertemente realacionada con otras, por lo que toda la información que recibe el modelo de esta la puede estar obteniendo, de forma indirecta con otras, haciendo que se vuelva prescindible.

## Tarea de Elección Abierta – Uso de SMOTE

Dado que el dataset está desbalanceado (con una mayoría de empleados que no abandonan la empresa), se ha optado por aplicar la técnica de sobremuestreo **SMOTE (Synthetic Minority Over-sampling Technique)**.

Esta técnica genera ejemplos sintéticos de la clase minoritaria (`Attrition = Yes`) para equilibrar el conjunto de datos durante el entrenamiento. El objetivo es que el modelo no tienda a favorecer a la clase mayoritaria, mejorando así su capacidad de detectar casos de abandono.

Se ha evaluado el mejor modelo anterior (**SVM ajustado con kernel='rbf' y C=10**) utilizando un pipeline que incorpora SMOTE y validación cruzada estratificada.

### Resultado obtenido:

- **Balanced Accuracy sin SMOTE**: **0.8756**
- **Balanced Accuracy media con SMOTE**: **0.8002**

Aunque el uso de **SMOTE** mejora el **equilibrio de las clases**, el **rendimiento global del modelo se ha reducido** al comparar la **balanced accuracy sin SMOTE** (0.8756) con la obtenida al aplicar **SMOTE** (0.8002). Este descenso en el rendimiento puede deberse a que el modelo ya estaba bastante bien ajustado sin la necesidad de equilibrar las clases artificialmente, lo que puede haber introducido ruido adicional al generar ejemplos sintéticos de la clase minoritaria.

Sin embargo, es importante destacar que SMOTE sigue siendo una técnica útil en **contextos de clases desbalanceadas**, ya que reduce el riesgo de que el modelo **ignore a la clase minoritaria**.

En resumen, **SMOTE** es una técnica que **puede ser útil** dependiendo del modelo, pero en este caso específico, su aplicación no ha mejorado el rendimiento en nuestro modelo.
