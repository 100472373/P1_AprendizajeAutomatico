## Importar Modelo Final

```python
import pandas as pd
import joblib

# Cargar el modelo entrenado desde el archivo
modelo_final = joblib.load("modelo_final.pkl")
```

## **Predicciones sobre Competition**

```python
# Cargamos los datos de competición
df_comp = pd.read_csv("attrition_competition_10.csv.gz")

# Si tiene ID, lo guardamos para incluirlo en la predicción
ids = df_comp["Id"] if "Id" in df_comp.columns else None

# Generamos predicciones
predicciones = modelo_final.predict(df_comp)

# Creamos el DataFrame con resultados
df_preds = pd.DataFrame({
    "Id": ids if ids is not None else range(len(predicciones)),
    "Attrition": predicciones
})
```

## Guardar Predicciones

```python
# Guardamos en CSV
df_preds.to_csv("predicciones.csv", index=False)
```
