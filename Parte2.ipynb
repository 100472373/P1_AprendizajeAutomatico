# **2. Decidir como se va realizar la evaluación**

El objetivo de esta sección es establecer una estrategia de evaluación fiable que nos permita:

- Comparar diferentes algoritmos y configuraciones (evaluación interna, o *inner*).
- Estimar el rendimiento futuro del modelo final en datos nuevos (evaluación externa, o *outer*).

Dado que se trata de un problema de clasificación **desbalanceado** (la variable objetivo `Attrition` no está equilibrada entre clases), hemos decidido usar métricas más apropiadas que el simple `accuracy`, como:

- **Balanced Accuracy**: media de TPR (True Positive Rate) y TNR (True Negative Rate).
- **TPR**: proporción de positivos correctamente clasificados.
- **TNR**: proporción de negativos correctamente clasificados.
- **Matriz de confusión**: para visualizar los aciertos y errores por clase.

La evaluación externa (*outer*) solo se utilizará **una vez**, al final del proceso, para estimar el rendimiento futuro. El resto de decisiones (modelo, escalado, hiperparámetros, etc.) se tomarán exclusivamente con evaluación interna (*inner*) sobre el conjunto de entrenamiento.

## Evaluación Outer, División en train/test

```python
df = pd.read_csv('/content/attrition_availabledata_10.csv.gz')

from sklearn.model_selection import train_test_split

# Variable objetivo y atributos
X = df.drop(columns=["Attrition"])  # Asegúrate de que 'Attrition' es tu variable objetivo
y = df["Attrition"]

# División 2/3 - 1/3, estratificada para mantener proporción de clases
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=1/3,
    stratify=y,
    random_state=42
)

print("Tamaño conjunto entrenamiento:", X_train.shape)
print("Tamaño conjunto test:", X_test.shape)
```

Se ha realizado una partición estratificada de los datos en 2/3 para entrenamiento y 1/3 para test. Esta evaluación solo se usará **una vez al final**, después de haber elegido el mejor modelo con evaluación interna. De esta manera, garantizamos una estimación honesta del rendimiento del modelo final sobre datos nuevos.


## Evaluación Inner  con validación cruzada

```python
from sklearn.model_selection import StratifiedKFold

# Definición de la validación cruzada (5 folds, estratificada)
cv_inner = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

Durante el desarrollo del modelo, se utilizará validación cruzada estratificada (5-fold). Esto permite:

- Comparar diferentes modelos (KNN, Árboles, SVM, etc.).
- Ajustar hiperparámetros de forma segura.
- Evitar el uso indebido del conjunto de test (evita *data leakage*).

Al mantener la proporción original de clases en cada fold, garantizamos una evaluación robusta incluso en un problema desbalanceado como este.

## Métricas elegidas

Como el problema presenta un desbalance de clases, no utilizamos `accuracy` como métrica principal. En su lugar, empleamos:

- `Balanced Accuracy`: más adecuada en este contexto, ya que exige buen rendimiento tanto para positivos como para negativos.
- `TPR` y `TNR`: permiten observar el rendimiento separado por clase.
- `Accuracy`: solo como referencia secundaria.
- `Matriz de confusión`: útil para analizar errores por tipo.

Estas métricas se utilizarán en las evaluaciones internas y, posteriormente, en la evaluación final del modelo sobre el conjunto de test.
